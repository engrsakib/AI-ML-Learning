## Ans to the question no 1

```python
import pandas as pd
import numpy as np

CSV_FILE = "final-employee-ds.csv"
COL = "TrainingHours"


df = pd.read_csv(CSV_FILE, usecols=[COL])
arr = pd.to_numeric(df[COL], errors="coerce").dropna().to_numpy(dtype=float)

mean = np.mean(arr)
std = np.std(arr)

print("mean", mean)
print("standard deviation", std)
```

## Ans to the question no 2

```python
SALARY_COL = "Salary"
DEPT_COL = "Department"
EXP_COL = "ExperienceYears"
PROJ_COL = "ProjectCount"

df = pd.read_csv(CSV_FILE)

df[SALARY_COL] = pd.to_numeric(df[SALARY_COL], errors="coerce")
df[EXP_COL] = pd.to_numeric(df[EXP_COL], errors="coerce")
df[PROJ_COL] = pd.to_numeric(df[PROJ_COL], errors="coerce")

dept_median = df.groupby(DEPT_COL)[SALARY_COL].transform("median")
overall_median = df[SALARY_COL].median()
df[SALARY_COL] = df[SALARY_COL].fillna(dept_median).fillna(overall_median)

mask = (df[EXP_COL] > 12) & (df[PROJ_COL] > 5)
total_salary = df.loc[mask, SALARY_COL].sum()
print("total slary",total_salary)
```

## Ans to the question no 3

```python
import pandas as pd

df = pd.read_csv(CSV_FILE)


df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["PerformanceScore"] = pd.to_numeric(df["PerformanceScore"], errors="coerce")


mask = (
    df["Department"].isin(["IT", "Finance"]) &
    df["Age"].between(30, 45) &
    (df["PerformanceScore"] > 88)
)

result = (
    df.loc[mask, ["Name", "Department", "Salary", "PerformanceScore"]]
    .dropna(subset=["Salary", "PerformanceScore"])
    .sort_values(by="Salary", ascending=False)
    .reset_index(drop=True)
)


if result.empty:
    print("No employees match the criteria.")
else:
    print(result.to_string(index=False))
```

## Ans to the question no 4

```python
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
df["WorkHoursPerWeek"] = pd.to_numeric(df["WorkHoursPerWeek"], errors="coerce")
df["SalaryPerHour"] = (df["Salary"] / (df["WorkHoursPerWeek"] * 4))

top_5 = df.nlargest(5, "SalaryPerHour")

ans = top_5[["Name", "SalaryPerHour"]].reset_index(drop=True)

print(ans.to_string(index=False))
```
